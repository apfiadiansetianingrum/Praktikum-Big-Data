```python
from pymongo import MongoClient

# ganti dengan connection string dari Atlas
client = MongoClient("mongodb+srv://apfiadian:apfdnsn22@cluster0.ulfohuq.mongodb.net/")

# pilih database sample_airbnb
db = client["sample_airbnb"]

# pilih collection listingsAndReviews
col = db["listingsAndReviews"]
```

```python
result = col.find({"room_type": "Entire home/apt"}).limit(10)  # limit biar gak kebanyakan
for doc in result:
    print(doc["name"], "-", doc["room_type"])
```

```python
result = col.find({"property_type": "House"}).limit(10)
for doc in result:
    print(doc["name"], "-", doc["property_type"])
```

```python
result = col.find({"bed_type": "Real Bed"}).limit(10)
for doc in result:
    print(doc["name"], "-", doc["bed_type"])
```

```python
result = col.find({"cancellation_policy": "flexible"}).limit(10)
for doc in result:
    print(doc["name"], "-", doc["cancellation_policy"])
```

```python
result = col.find({"minimum_nights": {"$gte": "2"}}).limit(10)
for doc in result:
    print(doc["name"], "-", doc["minimum_nights"], "malam")
```

```python
pipeline = [
    {"$addFields": {"minNightsInt": {"$toInt": "$minimum_nights"}}},
    {"$match": {"minNightsInt": {"$gte": 2}}},
    {"$limit": 10}
]

result = col.aggregate(pipeline)
for doc in result:
    print(doc["name"], "-", doc["minNightsInt"], "malam")
```
