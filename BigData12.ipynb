```python
#Exercise1
import pandas as pd

#Sample dataset with missing values
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', None],
    'Age': [24, 30, None, 22, 35],
    'Salary': [48000, None, 57000, None, 60000]
}

df = pd.DataFrame(data)
print("Original Data:\n", df)

#Filling missing values and dropping rows
df['Age'].fillna(df['Age'].mean(), inplace=True)
df['Salary'].fillna(df['Salary'].median(), inplace=True)
df.dropna(subset=['Name'], inplace=True)

print('\nAfter cleaning:\n', df)
```

```python
#Exercise1.1
import pandas as pd

#Sample dataset with missing values
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', None],
    'Age': [24, 30, None, 22, 35],
    'Salary': [48000, None, 57000, None, 60000]
}

df = pd.DataFrame(data)
print("Original Data:\n", df)

#Cara baru tanpa FutureWarning
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Salary'] = df['Salary'].fillna(df['Salary'].median())

#Drop row jika Name kosong
df = df.dropna(subset=['Name'])

print("\nAfter cleaning:\n", df)
```

```python
#Exercise2
import pandas as pd

#Sample dataset with inconsistent categorical values
data = {
    'Product': ['Laptop', 'Laptop', 'Desktop', 'Tablet', 'Tablet'],
    'Category': ['Electronics', 'electronics', 'Electronics', 'Gadgets', 'gadgets']
}

df = pd.DataFrame(data)
print("Original Data:\n", df)

#Standardize category values
df['Category'] = df['Category'].str.capitalize()

print('\nStandardized Data:\n', df)
```

```python
#Tugas
import pandas as pd
import numpy as np

#LOAD DATASET
#Membaca dataset film
df = pd.read_csv("movie_dataset.csv")

print("# Original Data Head")
print(df.head(), "\n")

print("# Missing Values per Kolom")
print(df.isnull().sum(), "\n")

#PENANGANAN MISSING VALUES
#Kolom numerik → isi dengan mean/median
df['runtime'] = df['runtime'].fillna(df['runtime'].mean())
df['budget']  = df['budget'].fillna(df['budget'].median())

#Kolom kategorikal → isi dengan placeholder agar tidak kosong
df['homepage'] = df['homepage'].fillna("Unknown")
df['overview'] = df['overview'].fillna("No overview")
df['tagline']  = df['tagline'].fillna("No tagline")
df['cast']     = df['cast'].fillna("Unknown")
df['director'] = df['director'].fillna("Unknown")
df['release_date'] = df['release_date'].fillna("Unknown")
df['genres']   = df['genres'].fillna("Unknown")
df['keywords'] = df['keywords'].fillna("No keywords")

#NORMALISASI KOLOM NUMERIK (Min–Max)
def minmax(col):
    return (col - col.min()) / (col.max() - col.min())

df['runtime_norm'] = minmax(df['runtime'])
df['budget_norm']  = minmax(df['budget'])
df['revenue_norm'] = minmax(df['revenue'])

#STANDARISASI KOLOM KATEGORIKAL
#Mengubah huruf agar seragam
df['original_language'] = df['original_language'].astype(str).str.lower()
df['status']            = df['status'].astype(str).str.capitalize()
df['title']             = df['title'].astype(str).str.strip()

#MENGHAPUS DUPLIKAT
before = df.shape[0]
df = df.drop_duplicates()
after = df.shape[0]

print("# Jumlah baris duplikat yang dihapus:", before - after, "\n")

#OUTPUT
print("# Cleaned Data Head")
print(df.head(), "\n")

print("# Missing Values Setelah Cleaning")
print(df.isnull().sum(), "\n")

print("# Final Data Shape:", df.shape)
```
