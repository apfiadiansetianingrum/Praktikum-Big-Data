```python
#tugas2
from pymongo import MongoClient

#koneksi ke MongoDB
client = MongoClient("mongodb://localhost:27017/")
db = client["company_db"]
collection = db["employees"]

#CREATE (Insert Data)
#insert satu data
employee_data = {
    "name": "Alice",
    "department": "Finance",
    "age": 29,
    "salary": 4500
}
collection.insert_one(employee_data)

#insert beberapa data tambahan
employees_bulk = [
    {"name": "Bob", "department": "IT", "age": 31, "salary": 5200},
    {"name": "Charlie", "department": "HR", "age": 26, "salary": 4000},
    {"name": "Diana", "department": "Marketing", "age": 28, "salary": 4800},
    {"name": "Evan", "department": "IT", "age": 35, "salary": 6000}
]
collection.insert_many(employees_bulk)

print("✅ Data berhasil disisipkan!")

#READ (Query Data)
print("\n📋 Data karyawan saat ini:")
for employee in collection.find():
    print(employee)

#UPDATE (Perbarui Data)
#contoh: update gaji Bob menjadi 5500
collection.update_one(
    {"name": "Bob"},
    {"$set": {"salary": 5500}}
)
print("\n📝 Gaji Bob berhasil diperbarui!")

#cek hasil update
updated_bob = collection.find_one({"name": "Bob"})
print("Data Bob setelah update:", updated_bob)

#DELETE (Hapus Data)
#contoh: hapus karyawan dengan nama Charlie
collection.delete_one({"name": "Charlie"})
print("\n❌ Data Charlie telah dihapus!")

#tampilkan data akhir
print("\n📊 Data karyawan setelah operasi CRUD:")
for employee in collection.find():
    print(employee)
```

```python
#tugas3
from pymongo import MongoClient

#koneksi ke database dan koleksi
client = MongoClient("mongodb://localhost:27017/")
db = client["company_db"]
collection = db["employees"]

#pipeline agregasi untuk menghitung rata-rata gaji per departemen
pipeline = [
    {
        '$group': {
            '_id': '$department',
            'average_salary': {'$avg': '$salary'}
        }
    },
    {
        '$sort': {'average_salary': -1}  #urutkan dari yang tertinggi
    }
]

#jalankan agregasi dan tampilkan hasilnya
print("Rata-rata gaji per departemen:\n")
for result in collection.aggregate(pipeline):
    print(f"Departemen: {result['_id']} | Rata-rata Gaji: {result['average_salary']:.2f}")
```

```python
#tugas5
from pymongo import MongoClient
import random
from datetime import datetime, timedelta

#koneksi ke MongoDB
client = MongoClient("mongodb://localhost:27017/")
db = client["company_db"]
employees = db["employees"]

#bersihkan koleksi lama
employees.delete_many({})

#membuat data 200 karyawan
departments = ["IT", "Finance", "HR", "Marketing", "Operations", "Logistics", "Design"]
positions = {
    "IT": ["Software Engineer", "System Analyst", "Network Admin", "Data Engineer"],
    "Finance": ["Accountant", "Financial Analyst", "Auditor"],
    "HR": ["HR Officer", "Recruiter", "HR Manager"],
    "Marketing": ["Marketing Specialist", "Content Creator", "SEO Analyst"],
    "Operations": ["Operations Staff", "Supervisor", "Coordinator"],
    "Logistics": ["Driver", "Warehouse Staff", "Logistic Manager"],
    "Design": ["Graphic Designer", "UI/UX Designer"]
}

first_names_male = ["Budi", "Andi", "Rizki", "Fajar", "Dimas", "Bagus", "Yusuf", "Arif", "Rendi", "Galang"]
first_names_female = ["Siti", "Dewi", "Ayu", "Nisa", "Putri", "Lia", "Rani", "Anisa", "Maya", "Indah"]
last_names = ["Santoso", "Wijaya", "Pratama", "Saputra", "Hidayat", "Permata", "Utami", "Rahmawati", "Anggraini", "Fauzi"]

data_bulk = []
for i in range(200):
    gender = random.choice(["Male", "Female"])
    if gender == "Male":
        name = f"{random.choice(first_names_male)} {random.choice(last_names)}"
    else:
        name = f"{random.choice(first_names_female)} {random.choice(last_names)}"

    dept = random.choice(departments)
    pos = random.choice(positions[dept])
    age = random.randint(22, 55)
    salary = random.randint(3500, 12000)
    hire_date = datetime.now() - timedelta(days=random.randint(100, 3650))  # 10 tahun terakhir

    employee = {
        "name": name,
        "gender": gender,
        "department": dept,
        "position": pos,
        "age": age,
        "salary": salary,
        "hire_date": hire_date.strftime("%Y-%m-%d")
    }
    data_bulk.append(employee)

employees.insert_many(data_bulk)
print("200 data karyawan berhasil dimasukkan ke koleksi 'employees'!\n")

#tugas 5.1 5 karyawan dengan gaji tertinggi per departemen
pipeline_top5 = [
    {"$sort": {"department": 1, "salary": -1}},
    {"$group": {
        "_id": "$department",
        "top_employees": {"$push": {"name": "$name", "salary": "$salary", "position": "$position"}}
    }},
    {"$project": {"top_5": {"$slice": ["$top_employees", 5]}}}
]

print("5 Karyawan dengan Gaji Tertinggi di Setiap Departemen:\n")
for dept in employees.aggregate(pipeline_top5):
    print(f"Departemen: {dept['_id']}")
    for emp in dept["top_5"]:
        print(f"   {emp['name']} - {emp['position']} | Gaji: {emp['salary']}")
    print()

#tugas 5.2 hapus karyawan dibawah umur 25 tahun
deleted = employees.delete_many({"age": {"$lt": 25}})
print(f" {deleted.deleted_count} karyawan di bawah usia 25 tahun telah dihapus!\n")

#tugas 5.3 laporan ringkas total gaji & rata-rata umur
pipeline_summary = [
    {
        "$group": {
            "_id": "$department",
            "total_salary": {"$sum": "$salary"},
            "average_age": {"$avg": "$age"},
            "employee_count": {"$sum": 1}
        }
    },
    {"$sort": {"_id": 1}}
]

print("Laporan Ringkas per Departemen (Setelah Penghapusan):\n")
for res in employees.aggregate(pipeline_summary):
    print(f"{res['_id']:<12} | Total Gaji: {res['total_salary']:<8} | "
          f"Rata-rata Umur: {res['average_age']:.2f} | Jumlah Karyawan: {res['employee_count']}")
```
