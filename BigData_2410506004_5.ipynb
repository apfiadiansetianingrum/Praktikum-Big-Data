```python
#tugas 1
#dataset sederhana
text = "Hadoop adalah sistem big data Hadoop memproses data dengan MapReduce"

#map function
def map_function(text):
    for word in text.split():
        yield (word, 1)

#reduce function
from collections import defaultdict

def reduce_function(pairs):
    result = defaultdict(int)
    for word, count in pairs:
        result[word] += count
    return result

#jalankan map
mapped = list(map_function(text))
print("Hasil Map (key-value pairs):")
print(mapped)

#jalankan reduce
reduced = reduce_function(mapped)
print("\nHasil Reduce (jumlah kata):")
print(dict(reduced))
```

```python
#tugas 2

dataset = """
MapReduce adalah model pemrograman untuk memproses data besar.
Map memecah data menjadi pasangan key-value.
Reduce menggabungkan data dan menghasilkan ringkasan.
"""

#jalankan map
mapped = list(map_function(dataset))
print("Hasil Map (key-value pairs):")
print(mapped[:15], "...")  # tampilkan 15 hasil pertama

#jalankan reduce
reduced = reduce_function(mapped)
print("\nHasil Reduce (jumlah kata):")
for word, count in reduced.items():
    print(f"{word}: {count}")
```
