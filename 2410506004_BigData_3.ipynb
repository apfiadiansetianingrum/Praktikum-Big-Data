```python
from pyspark.sql import SparkSession

#membuat SparkSession
spark = SparkSession.builder.appName("DataMahasiswa").getOrCreate()

#data mahasiswa
data = [
    ('Ayato', 'Informatika', 2021),
    ('Bennet', 'Sistem Informasi', 2020),
    ('Citlali', 'Informatika', 2022),
    ('Klee', 'Teknik Industri', 2021),
    ('Xiao', 'Sistem Informasi', 2022)
]

#nama kolom
columns = ['Nama', 'Jurusan', 'Angkatan']

#membuat dataframe
df = spark.createDataFrame(data, schema=columns)

print("DataFrame Mahasiswa:")
df.show()

print("Skema DataFrame:")
df.printSchema()

print("3 Baris Data Pertama:")
print(df.head(3))
```

```python
from pyspark.sql.functions import avg, max, min, count

#data nilai
data_nilai = [
    ('Ayato', 'Matematika', 85),
    ('Bennet', 'Matematika', 70),
    ('Citlali', 'Matematika', 90),
    ('Klee', 'Statistika', 75),
    ('Xiao', 'Statistika', 88),
    ('Ayato', 'Statistika', 92),
    ('Bennet', 'Pemrograman', 80),
    ('Citlali', 'Pemrograman', 95),
    ('Klee', 'Pemrograman', 78),
    ('Xiao', 'Pemrograman', 85)
]

columns_nilai = ['Nama', 'MataKuliah', 'Nilai']

#membuat dataframe
df_nilai = spark.createDataFrame(data_nilai, schema=columns_nilai)

print("DataFrame Nilai:")
df_nilai.show()

#1. pilih nama dan nilai
print("Nama dan Nilai:")
df_nilai.select('Nama', 'Nilai').show()

#2. filter nilai di atas 80
print("Mahasiswa dengan Nilai > 80:")
df_nilai.filter(df_nilai['Nilai'] > 80).show()

#3. rata-rata nilai per mata kuliah
print("Rata-rata Nilai per Mata Kuliah:")
df_nilai.groupBy('MataKuliah').avg('Nilai').show()

#4. nilai maksimum per mata kuliah
print("Nilai Maksimum per Mata Kuliah:")
df_nilai.groupBy('MataKuliah').max('Nilai').show()

#5.jumlah mahasiswa per mata kuliah
print("Jumlah Mahasiswa per Mata Kuliah:")
df_nilai.groupBy('MataKuliah').count().show()
```

```python
from pyspark.sql.functions import col

#data produk
data_produk = [
    ('Laptop', 'Elektronik', 8000),
    ('HP', 'Elektronik', 5000),
    ('Meja', 'Furniture', 2000),
    ('Kursi', 'Furniture', 1500)
]

columns_produk = ['NamaProduk', 'Kategori', 'Harga']

df_produk = spark.createDataFrame(data_produk, schema=columns_produk)

print("Data Produk:")
df_produk.show()

#menambahkan kolom diskon 10%
df_produk = df_produk.withColumn("Diskon10%", col("Harga") * 0.1)
df_produk.show()

#menambahkan kolom harga setelah diskon
df_produk = df_produk.withColumn("HargaFinal", col("Harga") - col("Diskon10%"))
df_produk.show()
```

```python
#data mahasiswa
data_mhs = [
    (1, 'Ayato', 'Informatika'),
    (2, 'Bennet', 'Sistem Informasi'),
    (3, 'Congyun', 'Teknik Industri')
]
columns_mhs = ['NPM', 'Nama', 'Jurusan']
df_mhs = spark.createDataFrame(data_mhs, schema=columns_mhs)

#data nilai
data_nilai = [
    (1, 'Matematika', 85),
    (2, 'Matematika', 75),
    (3, 'Matematika', 90),
    (1, 'Pemrograman', 95),
    (2, 'Pemrograman', 80)
]
columns_nilai = ['NPM', 'MataKuliah', 'Nilai']
df_nilai = spark.createDataFrame(data_nilai, schema=columns_nilai)

print("Data Mahasiswa:")
df_mhs.show()
print("Data Nilai:")
df_nilai.show()

#inner join berdasarkan npm
print("Hasil Inner Join:")
df_join = df_mhs.join(df_nilai, on="NPM", how="inner")
df_join.show()

#left join (menampilkan semua mahasiswa walaupun tidak ada nilai)
print("Hasil Left Join:")
df_mhs.join(df_nilai, on="NPM", how="left").show()
```

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql.window import Window

#inisialisasi SparkSession
spark = SparkSession.builder.appName("CarSalesExploration").getOrCreate()

#load dataset
csv_file = path + "/car_sales_data.csv"
df = spark.read.csv(csv_file, header=True, inferSchema=True)

#data awal
print("Data Awal (5 baris)")
df.show(5)
df.printSchema()

#transformasi dasar
print("Pilih kolom Manufacturer & Price")
df.select("Manufacturer", "Price").show(5)

print("Filter mobil dengan harga > 20000")
df.filter(df["Price"] > 20000).show(5)

print("Rata-rata harga mobil per Manufacturer")
df.groupBy("Manufacturer").avg("Price").show()

#kolom baru
print("Tambah kolom Discount & Final Price")
df2 = df.withColumn("Discount", df["Price"] * 0.1)
df2 = df2.withColumn("FinalPrice", df2["Price"] - df2["Discount"])
df2.select("Manufacturer", "Model", "Price", "Discount", "FinalPrice").show(5)

#operasi lanjutan
print("Ranking mobil berdasarkan harga dalam tiap Manufacturer")
windowSpec = Window.partitionBy("Manufacturer").orderBy(df["Price"].desc())
df_ranked = df.withColumn("Rank_in_Brand", F.rank().over(windowSpec))
df_ranked.select("Manufacturer", "Model", "Price", "Rank_in_Brand").show(10)

print("Running total harga berdasarkan Manufacturer")
windowSpec2 = Window.partitionBy("Manufacturer").orderBy("Price")\
                    .rowsBetween(Window.unboundedPreceding, Window.currentRow)
df_running = df.withColumn("RunningTotal", F.sum("Price").over(windowSpec2))
df_running.select("Manufacturer", "Model", "Price", "RunningTotal").show(10)
```
